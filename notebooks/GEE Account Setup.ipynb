## Before you start

Complete these steps **before the workshop**. Registration is reviewed by Google and approval is not always immediate, so do not leave it to the last minute.

You will need a **Google account** (Gmail). Institutional accounts sometimes block third-party sign-in, so have a personal account available as a backup.

> ⚠️ At **Step 3** you will choose a **Project ID**. Make a note of it — you will use it throughout the workshop to initialise Earth Engine.

## What is Google Earth Engine?

**Google Earth Engine (GEE)** is a cloud-based platform for geospatial analysis. It provides access to a large catalog of satellite imagery and other geospatial datasets, together with the computing power to process them.

- **Access to a massive data catalog:** petabytes of satellite imagery and other geospatial datasets
- **Scalable computing power:** Google's infrastructure for complex geospatial computations
- **Integration with Python libraries:** combine GEE with common data science packages
- **No large downloads:** analysis runs in the cloud

To learn more: https://earthengine.google.com/

## Step 1. Register a GEE account

Go to **https://code.earthengine.google.com/register** and, if prompted, sign in with your Google account.

You should land on a page titled *"Welcome to Earth Engine"*. Under **Get started**, click **Register**.

Welcome to Earth Engine overview page

## Step 2. Register your Cloud project

In the **Configuration** tab, click **Create project** to start configuring your project.

Earth Engine Configuration tab with the Create project button

## Step 3. Create the new project

Fill in the **New Project** form:

- **Project name** — type a name for your project.
- **Project ID** — 📝 **make a note of this.** It cannot be changed later, and you will use it throughout the workshop.
- **Location** — select **No organization**.

Then click **Create**.

New Project form showing project name, project ID and location

## Step 4. Fill out the registration form

The form has five numbered sections. Work through them in order.

**1. Select your organization type** — choose the option that best describes you, then click **Next**.

Step 1 of the form: select your organization type

**2. Check non-commercial eligibility** — select *Individual research or non-commercial use*, describe your use (for example, *Attending Workshop on GEE*), then click **Check eligibility**.

Step 2 of the form: check non-commercial eligibility

**3. Choose your plan** — a pricing plan is **not** required for non-commercial registration. Click **Next**.

Step 3 of the form: choose your plan

**4. Describe your work** — answer the questions about how you will use Earth Engine, then click **Next**.

Step 4 of the form: describe your work

**5. Review summary** — check your answers and click **Register**.

## Step 5. Enable the Earth Engine API

To complete registration, you are asked to enable the Earth Engine API for your project. Click **Enable**.

Enable required APIs dialog

## Great, you are done! 🎉

Your Google Earth Engine account and Cloud project are ready.
